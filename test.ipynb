```python
names = ['aa','bb','cc','dd','ee']

ladder_participate_count = len(names)

seperator = " " * ladder_participate_count
print((" " * (ladder_participate_count-1)).join(names))

'''
2. 리스트가 필요한 이유
리스트는 각 칸의 내용을 저장하는 데 사용됩니다. 한 줄에 있는 모든 칸의 데이터를 저장하기 위해 리스트가 필요합니다.

3. join()과 리스트의 관계
join()은 문자열 메서드로, 리스트에 저장된 문자열들을 특정 구분자(여기서는 |)로 연결해 하나의 문자열로 만듭니다.
'''
for i in range (ladder_participate_count):
        segments = [seperator if random.choice([0,ladder_participate_count]) == 0 else "-"*ladder_participate_count for _ in range(ladder_participate_count-1)]
        print("|" + "|".join(segments) + "|")
```
